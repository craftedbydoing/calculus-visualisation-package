### Import

```python
import numpy as np
from calc_package.core.sequences import sample_sequence, find_N_for_limit
from calc_package.plotting.sequences import plot_sequence, plot_convergence
from calc_package.widgets.sequences import sequence_convergence_explorer
import matplotlib.pyplot as plt
```

### Basic seq work

```python
n, a = sample_sequence(lambda n: 3-1/n, N=20)
fig, ax = plt.subplots()

plot_sequence(ax, n, a)
```

```python
n, a = sample_sequence(lambda n: 1/n)
L = 0
eps = 0.2

fig, ax = plt.subplots()

ax.axhline(L, color="black")              # limit line
ax.axhline(L + eps, linestyle="--")       # upper band line
ax.axhline(L - eps, linestyle="--")       # bottom band line
ax.axhspan(L - eps, L + eps, alpha=0.2)   # band
plot_sequence(ax, n, a)

inside = np.abs(a - L) < eps
outside = ~inside
print(inside)
print(outside)
print(np.where(outside)) # finds indexes, see the output (smth, )
print(np.where(outside)[0][-1])
```

Find N for limit

```python
n, a = sample_sequence(lambda n: 1/n)
L = 0
eps = 0.2

fig, ax = plt.subplots()

ax.axhline(L, color="black")              # limit line
ax.axhline(L + eps, linestyle="--")       # upper band line
ax.axhline(L - eps, linestyle="--")       # bottom band line
ax.axhspan(L - eps, L + eps, alpha=0.2)   # band
plot_sequence(ax, n, a)

print(find_N_for_limit(n, a, L, eps))
```

plot_convergence work

```python
n, a = sample_sequence(lambda n: 1/n)
L = 0
eps = 0.1
N = find_N_for_limit(n, a, L, eps)

fig, ax = plt.subplots()
plot_convergence(ax, n, a, L, eps, N)
```
