### Widgets work

```python
%matplotlib widget
functions = {
    "1/x":      lambda x: 1/x,
    "tan":      lambda x: np.tan(x),
    "1/(x-1)":  lambda x: 1/(x - 1),
}

fig, ax = plt.subplots(figsize= (6,6))

@interact(factor=(2, 200), num_points=(50, 2000), func_names=list(functions.keys()))
def show(factor, num_points, func_names):
    ax.clear() # clears the fig
    x, y = f.sample_clean(functions[func_names], (-5, 5), num_points=num_points, factor=factor)
    plot.plot_function(ax, x, y)
    fig.canvas.draw_idle() # "redraw command"
```
