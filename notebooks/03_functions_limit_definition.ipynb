### Eps-delta definition craft

```python
x, y = sample_clean(lambda x: np.sin(x)/x, (-np.pi/2,np.pi/2))
fig, ax = plt.subplots()

L = 1
eps = 0.3
plot_function(ax, x, y)
ax.axhline(L, color="black")
ax.axhline(L + eps, linestyle="--")
ax.axhline(L - eps, linestyle="--")
ax.axhspan(L - eps, L + eps, alpha=0.2)

a = 0
# find delta
inside_mask = np.abs(y - L) < eps
# inside_mask[220:230]
outside_idx = np.where(~inside_mask)[0] # indexes outside, output (smth, )
# print(outside_idx)

# if len(outside_idx) == 0:
#     return n[0] # all elements inside band
# if len(outside_idx) == len(a): # no evidence sequence converges yet
#     return None

x_idxs_right_of_a = np.where(x > a)[0]
x_idxs_left_of_a = np.where(x < a)[0]
del_right_idx = np.min(np.intersect1d(x_idxs_right_of_a, outside_idx))
del_left_idx = np.max(np.intersect1d(x_idxs_left_of_a, outside_idx))

print(min(abs(x[del_left_idx] - a), abs(x[del_right_idx] - a)))
# print(last_out)
```

```python
x, y = sample_clean(lambda x: x**2, (-1,3))
a = 2
L = 4
eps = 0.7
delta = find_delta_for_limit(x,y,a,L,eps)
print(delta)
fig, ax = plt.subplots()
plot_function(ax, x, y)
ax.axhline(L, color="black")
ax.axhline(L + eps, linestyle="--")
ax.axhline(L - eps, linestyle="--")
ax.axhspan(L - eps, L + eps, alpha=0.2)
ax.axvline(a + delta, linestyle="--", color="green")
ax.axvline(a - delta, linestyle="--", color="green")
```
