Derivative error using numerics

```python
f = lambda x: np.exp(x)
a = 0.0
actuall_derivative = np.exp(a)

h_vals = np.logspace(-16, 0, 100)
numeric_derivative = [derivative_at_point(f, a, h) for h in h_vals]
err = [abs(actuall_derivative - element) for element in numeric_derivative]

fig, ax = plt.subplots()
ax.loglog(h_vals, err)
ax.set_xlabel("h")
ax.set_ylabel("|error|")
ax.set_title(f"Central difference error for $e^x$ at {a}")
ax.grid(True, which="both", alpha=0.3)
```

```python
f = lambda x: np.sin(x)
a = 1.0
actuall_derivative = np.cos(a)

h_vals = np.logspace(-16, 0, 100)
numeric_derivative = [derivative_at_point(f, a, h) for h in h_vals]
err = [abs(actuall_derivative - element) for element in numeric_derivative]

fig, ax = plt.subplots()
ax.loglog(h_vals, err)
ax.set_xlabel("h")
ax.set_ylabel("|error|")
ax.set_title(f"Central difference error for $sin(x)$ at {a}")
ax.grid(True, which="both", alpha=0.3)
```

Comparison

```python
f = lambda x: np.exp(x)
a = 0.0
actuall_derivative = np.exp(a)

err_central = [abs(actuall_derivative - derivative_at_point(f, a, h)) for h in h_vals]
err_forward = [abs(actuall_derivative - calculate_line_slope(f, a, h)) for h in h_vals]

fig, ax = plt.subplots()
ax.loglog(h_vals, err_central, label="central der")
ax.loglog(h_vals, err_forward, label="forward der")
ax.set_xlabel("h")
ax.set_ylabel("|error|")
ax.set_title(f"Numerical derivative error for $e^x$ at {a}")
ax.grid(True, which="both", alpha=0.3)
ax.legend();
```
